# Building a Regression MLP sing the Sequential API

```python
from sklearn.datasets import fetch_california_housing
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
import tensorflow as tf
from tensorflow import keras
import pandas as pd
import numpy as np
```

First, let's download the California Housing dataset from a direct URL. We'll use `wget` for this.

```python
import requests

url = "https://raw.githubusercontent.com/ageron/handson-ml2/master/datasets/housing/housing.csv"
output_path = "housing.csv"

response = requests.get(url)
response.raise_for_status()  # Raise an exception for HTTP errors

with open(output_path, "wb") as f:
    f.write(response.content)

print(f"Downloaded {output_path}")
```

```python
housing = pd.read_csv('housing.csv')
housing.head()
```

```python
housing.drop(columns="ocean_proximity", axis=1, inplace=True)
```

```python
housing.isnull().sum()
```

```python
housing.dropna(axis=0, inplace=True)
```

```python
housing.duplicated().sum()
```

```python
housing.head()
```

Now that the `housing.csv` file is downloaded, we can load it into a pandas DataFrame.

## Train Test Split

```python
X_train_full, X_test,y_train_full, y_test = train_test_split(housing.drop('median_house_value', axis=1), housing['median_house_value'], random_state=42)
X_train, X_valid, y_train, y_valid = train_test_split(X_train_full, y_train_full)
```

```python
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_valid = scaler.transform(X_valid)
X_test = scaler.transform(X_test)
```

```python
scaler_y = StandardScaler()
y_train = scaler_y.fit_transform(np.array(y_train).reshape(-1, 1))
y_valid = scaler_y.transform(np.array(y_valid).reshape(-1, 1))
y_test = scaler_y.transform(np.array(y_test).reshape(-1, 1))
```

## Using the Sequential API To Build, Train and Evaluate

```python
model = keras.models.Sequential([
    keras.layers.Dense(30, activation="relu", input_shape=X_train.shape[1:]),
    keras.layers.Dense(1)
])
```

```python
from tensorflow.keras.optimizers import Adam

# Default is usually 0.001. Try 0.0001 or 0.00001
opt = Adam(learning_rate=0.0001)
```

```python
model.compile(loss="mse", optimizer=opt, metrics=["mse"])
history = model.fit(X_train, y_train, epochs=20, validation_data=(X_valid, y_valid))
```

```python
mse_test = model.evaluate(X_test, y_test)
```

```python
X_new = X_test[:3]
y_pred = model.predict(X_new)
```
