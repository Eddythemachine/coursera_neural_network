Converting RGB Image to Greyscale Image

```python
img = cv.imread("dog.jpg")
```

```python
img
```

```python
type(img)
```

```python
greyscale = cv.cvtColor(img, cv.COLOR_BGR2GRAY)
```

```python
greyscale.shape
```

```python
plt.imshow(greyscale, cmap="gray")
plt.show()
```
